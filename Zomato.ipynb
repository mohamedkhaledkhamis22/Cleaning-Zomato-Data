# Data Cleaning:

Deleting redundant columns.

Renaming the columns.

Dropping duplicates.

Cleaning individual columns.

Remove the NaN values from the dataset

Check for some more Transformations


```python
import pandas as pd
```

```python
df=pd.read_csv("zomato.csv")
```

```python
df.head(5)
```

```python
#df.tail(10)
```

```python
df.columns
```

```python
#Step1 : Deleting redundant columns
redundant_columns_drop=['url', 'address','reviews_list', 'menu_item',
       'listed_in(type)', 'listed_in(city)']
df.drop(columns=(redundant_columns_drop),inplace=True)
```

```python
df
```

```python
df.columns
```

```python
#Step2 : Renaming the columns
df.rename(columns={'approx_cost(for two people)':'cost for two'},inplace=True)
```

```python
df
```

```python
#Step3 : Dropping duplicates
df.duplicated().sum()
```

```python
df.drop_duplicates()
```

```python
#Step 4 : Cleaning individual columns
#Rate Coulmns
df['rate']=df['rate'].str.replace('/',' out of ')
```

```python
df
```

```python
#Name
df['name']=df['name'].str.lower()
```

```python
df
```

```python
#Step 5 : Remove the NaN values from the dataset
df.isna().sum()
```

```python
df=df.dropna()
```

```python
df
```

```python
#Step 6 : Check for some more Transformations
df['cost for two'] = df['cost for two'].astype(str).str.replace(',','', regex=True)
df['cost for two'] = pd.to_numeric(df['cost for two'], errors='coerce')
```

```python
df['Cost For Person'] =df['cost for two']*0.5
```

```python
df
```

```python
df.to_csv("Cleaned Zomato.csv",index=False)
```
